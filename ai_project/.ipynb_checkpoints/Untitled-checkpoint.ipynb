```python
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from scipy.sparse import hstack

# 1. Load dataset
df = pd.read_csv(r'C:\Users\maryu\Documents\AI Project\Dataset\sleep_disorder_dataset_1.csv')

# 2. Encode categorical data
le_gender = LabelEncoder()
df['Gender_Encoded'] = le_gender.fit_transform(df['Gender'])

le_target = LabelEncoder()
y = le_target.fit_transform(df['Sleep_Disorder_Type'])

# 3. Vectorize text (Symptoms)
tfidf = TfidfVectorizer(stop_words='english', max_features=1500)
X_text = tfidf.fit_transform(df['OCR_Extracted_Text'])

# 4. Combine Features
X_clinical = df[['Age', 'Gender_Encoded', 'AHI_Score', 'SaO2_Level', 'Diagnosis_Confirmed']].values
X_combined = hstack([X_clinical, X_text])

# 5. Train Model
rf_model = RandomForestClassifier(n_estimators=200, random_state=42)
rf_model.fit(X_combined, y)

# 6. Save components
joblib.dump(rf_model, 'model.pkl')
joblib.dump(tfidf, 'tfidf.pkl')
joblib.dump(le_target, 'target_encoder.pkl')
joblib.dump(le_gender, 'gender_encoder.pkl')

print("AI Brain Ready: Fully trained on data!")
 
```
